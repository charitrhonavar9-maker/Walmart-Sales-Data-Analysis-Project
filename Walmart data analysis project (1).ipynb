# Solving business problems
### Q.1 Find the different payment method and number of transaction, number of qty sold. ###


```python
query = """

SELECT payment_method, COUNT(*) as no_payments, sum(quantity) as no_qty_sold
from walmart_raw
GROUP BY payment_method

"""

df = pd.read_sql(query, engine)
df
```

### Q.2 identify the highest-rated category in each branch, displaying the branch, category and Avg rating.


```python
query = """

SELECT branch, category,  AVG(rating) as avg_rating,
RANK() OVER(PARTITION BY branch ORDER BY AVG(rating) DESC) as ranks
FrOM walmart_raw
GROUP BY 1, 2

"""
df = pd.read_sql(query, engine)
df
```

### Q.2.a Selecting only rank 1


```python
query = """

SELECT  *
From 
(
	SELECT 
      branch, 
      category,  
      AVG(rating) as avg_rating,
      RANK() OVER(PARTITION BY branch ORDER BY AVG(rating) DESC) as ranks
      From walmart_raw
      GROUP BY 1, 2
) as ranked_data

WHERE ranks = 1

"""
df = pd.read_sql(query, engine)
df
```

### Q.3 Identfying the busiest day for each branch on the number of transaction.

```python
query = """

SELECT *
FROM
(
SELECT branch,
dayname(str_to_date(date, '%%d/%%m/%%y')) as day,
COUNT(*) as no_transaction,
RANK() OVER(PARTITION by branch ORDER  BY COUNT(*) DESC) as ranks
FROM walmart_raw
GROUP BY 1, 2
) as ranks
WHERE ranks = 1

"""

df = pd.read_sql(query, engine)
df
```

### Q.4 Calculate the total number of quantity items sold per payment method. List the payment method and total quantity.

```python
query = """

SELECT payment_method, COUNT(*) as no_payments, sum(quantity) as no_qty_sold
from walmart_raw
GROUP BY payment_method

"""

df = pd.read_sql(query, engine)
df
```

###   Q.5 Determine the average, minimum, maximum rating of products for each city. List the city, avrage_rating, min_rating and max_rating.

```python
query = """

SELECT city, category,
AVG(rating) as avrage_rating, min(rating) as min_rating, max(rating) as max_rating
from walmart_raw
GROUP BY 1, 2
ORDER BY 1, 3, 4, 5

"""

df = pd.read_sql(query, engine)
df
```

### Q.6 Calculate the total profit for each category by considering total profit as (unit_price * quanity * profit_margin ) 
###     List the category and total_profit, ordered from highest to lowest profit.

```python
query = """

SELECT category, sum(unit_price * quantity * profit_margin) as total_profit
FROM walmart_raw
GROUP BY 1
ORDER BY total_profit DESC

"""

df = pd.read_sql(query, engine)
df
```

### Q.7 Determine the most common payment method for each branch. Display branch and the preffered_payment_method

```python
query = """

WITH cte as 
(
SELECT branch, payment_method, COUNT(*) as no_payments,
RANK() OVER(PARTITION BY branch ORDER BY COUNT(*) DESC) as  ranks
from walmart_raw
GROUP BY 1, 2
)
SELECT *
FROM cte
where ranks = 1

"""


df = pd.read_sql(query, engine)
df


```

### Q.8 categorize sales into 3 groups MORNING, AFTERNOON, EVNING. FIND out which of the shift and number of invoice

```python
query = """

SELECT  branch,  
CASE 
   WHEN extract(HOUR FROM time) < 12 THEN 'MORNING'
   WHEN EXTRACT(HOUR FROM time) BETWEEN 12 AND 17 THEN 'AFTERNOON'
   ELSE 'EVENING'
END as shift, COUNT(*)
FROM walmart_raw
GROUP BY 1, 2
ORDER BY 1, 3 DESC

"""

df = pd.read_sql(query, engine)
df
```

### Q.9  Identify the 5 branch with the highest decrease ratio in revenue compare to last year current year 2023 and last year 2022 
###      revenue decrease ratio as rdr = ((last year revnue - current year revenue) / last year revneue) * 100

```python
query = """

WITH revenue_2022 AS
(
SELECT 
      branch, 
      sum(total_price) as revenue
from walmart_raw
WHERE year(str_to_date(date, '%%d/%%m/%%y')) = 2022
GROUP BY 1
),
revenue_2023 AS
(
SELECT 
      branch, 
      sum(total_price) as revenue
from walmart_raw
WHERE year(str_to_date(date, '%%d/%%m/%%y')) = 2023
GROUP BY 1
ORDER BY 1
)
SELECT ls.branch, ls.revenue as last_year_revenue, cs.revenue as current_year_revenue,
round((ls.revenue - cs.revenue)/ls.revenue * 100,2) as rdr
FROM revenue_2022 as ls
jOIN 
revenue_2023 as cs
ON ls.branch = cs.branch
WHERE ls.revenue > cs.revenue
ORDER BY 4 desc
LIMIT 5

"""

df = pd.read_sql(query, engine)
df
```
